## Bivariate Analysis

```python
sns.countplot(data=df,x="Gender",hue="Showed_up",color="Blue")
plt.title("Gender vs Showed_up")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

- The bar chart shows the distribution of male and female patients who attended and missed their appointments. It shows that female patients had a higher number of appointments than male patients. Among both genders, the majority of patients attended their appointments, while a smaller number missed them. Overall, attendance was higher than non-attendance for both male and female patients. 

```python
sns.countplot(data=df,x="Scholarship",hue="Showed_up",color="White",edgecolor="Black")
plt.title("Scholarship vs Showed_up")
plt.xlabel("Scholarship")
plt.ylabel("Count")
plt.show()
        
```

- The bar chart compares appointment attendance between patients with and without scholarships. It shows that most patients did not have a scholarship, and among them, the majority attended their appointments. Patients with scholarships were fewer in number, and most of them also attended their appointments. Overall, appointment attendance was higher than missed appointments in both scholarship groups.


```python
sns.countplot(data=df,x="Hypertension",hue="Showed_up",color="Red")
plt.title("Hypertension vs Showed_up")
plt.xlabel("Hypertension")
plt.ylabel("Count")
plt.show()
```

- The barchart compares appointment attendence with the patients having hypertension or not. Most of the patients are not having hypertension and majority of them have attended their appointment. Patients with hypertension were fewer in number, among them most of them had appeared. Overall, attendence was higher than missed appointments in both groups. 

```python
sns.countplot(data=df,x="Diabetes",hue="Showed_up",color="Pink")
plt.title("Diabetes vs Showed_up")
plt.xlabel("Diabetes")
plt.ylabel("Count")
plt.show()
```

- The bar chart compares the patients attendence between the patients with and without diabetes. Patients with diabetes is less than without diabetes. In both groups, their attendence is higher than their absence.

```python
sns.countplot(data=df,x="Alcoholism",hue="Showed_up",color="Yellow")
plt.title("Alcoholism vs Showed_up")
plt.xlabel("Alcoholism")
plt.ylabel("Count")
plt.show()
```

- This bar chart is a comparison between the patients using alcohols and not with their appearance for the appointment. Most of them are not using alcohols. The attendence of the patients is higher in both groups. 

```python
sns.countplot(data=df,x="Handicap",hue="Showed_up",color="Green")
plt.title("Handicapped vs Showed_up")
plt.xlabel("Handicap")
plt.ylabel("Count")
plt.show()
```

- The bar chart compares appointment attendance between patients with and without disabilities. ## Multivariate Analysis

```python
sns.catplot(data=df,x="Gender",hue="Showed_up",col="Scholarship",kind="count",color="Red",edgecolor="Black")
plt.title("Gender and Scholarship vs Showed_up")
plt.show()
```

- In both scholarship groups, female patients had more appointments than male patients. The majority of both male and female patients attended their appointments, while fewer patients missed their appointments. Additionally, the number of patients without scholarships was much higher than the number of patients with scholarships.

```python
sns.catplot(data=df,x="Gender",hue="Showed_up",col="Hypertension",kind="count",color="Blue",edgecolor="Black")
plt.title("Gender and Hypertension vs Showed_up")
plt.show()
```

- In both groups, female patients had more appointments than male patients. The majority of both male and female patients attended their appointments, while fewer patients missed their appointments. Overall, patients without hypertension were more numerous than patients with hypertension, and attendance was higher than missed appointments in both groups.



```python
sns.catplot(data=df,x="Gender",hue="Showed_up",col="Diabetes",kind="count",color="Brown",edgecolor="Black")
plt.title("Gender and Diabetes vs Showed_up")
plt.show()
```

-  In both groups, female patients had more appointments than male patients. The majority of both male and female patients attended their appointments, while fewer patients missed their appointments. Overall, patients without diabetes were more numerous than patients with diabetes, and attendance was higher than missed appointments in both groups.

# Conclusion

- This project analyzed the Medical Appointment No-Show dataset to identify the factors associated with patient appointment attendance. Data cleaning and exploratory data analysis (univariate, bivariate, and multivariate analyses) were performed to understand the dataset. The results showed that most patients attended their appointments, while a smaller proportion missed them. Female patients had more appointments than male patients. Most patients did not have scholarships, hypertension, diabetes, alcoholism, or disabilities. Across all these groups, the number of patients who attended appointments was consistently higher than those who missed them. The analysis also showed that patients without these conditions made up the majority of the dataset. Overall, the findings provide useful insights into patient attendance patterns and can help healthcare providers improve appointment scheduling, reminder systems, and strategies to reduce missed appointments.

# Recommendations
